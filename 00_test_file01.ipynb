```python
import pandas as pd
from pathlib import Path

types = ['logon', 'file', 'http', 'email', 'device']
for t in types:
    count = 0
    path = Path(f"output/{t}")
    if path.exists():
        for p in path.glob("*.parquet"):
            df = pd.read_parquet(p)
            count += df['is_anomaly'].sum()
    print(f"Tổng số nhãn 1 trong {t}: {count}")
```

```python
import pandas as pd
from pathlib import Path

# Các thư mục parquet
types = ['logon', 'file', 'http', 'email', 'device']

for t in types:
    print("\n" + "=" * 80)
    print(f"[{t.upper()}]")
    print("=" * 80)

    path = Path(f"output/{t}")

    if not path.exists():
        print("Không tồn tại thư mục")
        continue

    parquet_files = sorted(path.glob("*.parquet"))

    if not parquet_files:
        print("Không có file parquet")
        continue

    # đọc file đầu tiên để xem schema
    sample_file = parquet_files[0]

    print(f"\nSample file: {sample_file.name}")

    df = pd.read_parquet(sample_file)

    # ===== THÔNG TIN CƠ BẢN =====
    print(f"\nShape: {df.shape}")

    print("\nColumns:")
    for c in df.columns:
        print(f" - {c}")

    # # ===== KIỂU DỮ LIỆU =====
    # print("\nDtypes:")
    # print(df.dtypes)

    # # ===== THIẾU DỮ LIỆU =====
    # print("\nMissing values:")
    # print(df.isnull().sum())

    # # ===== THỐNG KÊ NHÃN =====
    # if "is_anomaly" in df.columns:
    #     print("\nLabel distribution:")
    #     print(df["is_anomaly"].value_counts(dropna=False))

    # # ===== XEM MẪU =====
    # print("\nHead:")
    # print(df.head(3))

    # ===== MEMORY =====
    mem_mb = df.memory_usage(deep=True).sum() / 1024 / 1024
    print(f"\nMemory usage: {mem_mb:.2f} MB")
```
